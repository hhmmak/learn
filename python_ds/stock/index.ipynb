```python
import pandas as pd
import numpy as np
import math
import datetime as dt
```

```python
aapl = pd.read_csv("./data/AAPL.csv", index_col="Date", parse_dates=True, usecols=["Date", "Adj Close"], na_values=["nan"])
ibm = pd.read_csv("./data/IBM.csv", index_col="Date", parse_dates=True, usecols=["Date", "Adj Close"], na_values=["nan"])
jpm = pd.read_csv("./data/JPM.csv", index_col="Date", parse_dates=True, usecols=["Date", "Adj Close"], na_values=["nan"])
spy = pd.read_csv("./data/SPY.csv", index_col="Date", parse_dates=True, usecols=["Date", "Adj Close"], na_values=["nan"])
aapl = aapl.rename(columns={"Adj Close": "AAPL"})
ibm = ibm.rename(columns={"Adj Close": "IBM"})
jpm = jpm.rename(columns={"Adj Close": "JPM"})
spy = spy.rename(columns={"Adj Close": "SPY"})
prices = pd.concat([aapl, ibm, jpm, spy], axis=1)
prices.dropna(subset=["SPY"], inplace=True)
prices
```

```python
# calc sma-14 for entire date range

# clear out df 
sma = pd.DataFrame(0,index=prices.index, columns=prices.columns)
sma
```

```python
#loop over all days
lookback = 14
for day in range(prices.shape[0]):

    # set to na if day too early for SMA
    if day < lookback:
        sma.iloc[day] = np.nan
        # sma[sma.isna().any(axis=1)] result in 14 rows
    else:

        for prev_day in range(day-lookback+1, day+1):
            for sym in range(4):
                sma.iloc[day,sym] = sma.iloc[day,sym] + prices.iloc[prev_day, sym]
        #     sma.iloc[day] = sma.iloc[day] + prices.iloc[prev_day]
        # sma.iloc[day] = prices.iloc[day-lookback+1 : day+1].sum()
        sma.iloc[day] = sma.iloc[day] / lookback
    
```
