```python
import numpy as np

# Example data
output = np.array([1, 2, 3, 4])  # 1D output array with 4 samples
# features = np.array([[1, 5],     # Feature matrix with 4 samples and 2 features
#                      [2, 4],
#                      [3, 3],
#                      [4, 2]])

features = np.array([[1, 5, 4],     # Feature matrix with 4 samples and 2 features
                     [2, 4, 4],
                     [3, 3, 3],
                     [4, 2, 3]])

# Concatenate the output as a new column in the features matrix

# Compute the correlation matrix
correlation_matrix = np.corrcoef(features, output, rowvar=False)

# Extract the correlations between output (first row/col) and the features
correlations_with_output = correlation_matrix
print(correlations_with_output)
```

```python

correlations_with_output = correlation_matrix[0, 1:]  # The first row (output) with all other features

print(correlations_with_output)
```

```python
import pandas as pd

# Create the original DataFrame
df = pd.DataFrame({
    "Col1": [0, 20, 0, 0, -20],
    "Col2": [-7, 0, 0, 3, 0],
    "Col3": [7, -20, 0, -3, 20]
}, index=pd.date_range("2020-01-01", "2020-01-05"))

# Create df3 and set the initial value
df3 = pd.DataFrame(0, index=df.index, columns=df.columns)
df3.at["2020-01-01", "Col3"] = 100
df3.iloc[0] = df.iloc[0] + df3.iloc[0]
# Calculate the values for df3 row by row
for i in range(1, len(df3)):
    df3.iloc[i] = df.iloc[i] + df3.iloc[i-1]

print(df3)
```
